# Gráficas para corroborar la interpretación de los resultados obtenidos por el modelo de simulacion.


```python
import pandas as pd

# Datos de las simulaciones
data = [
    [2, 0.5, 43.17, 80.23, 101.09],
    [2, 0.333, 42.11, 67.95, 100.69],
    [2, 0.2, 32.43, 38.33, 98.18],
    [4, 0.5, 35.82, 55.60, 100.70],
    [4, 0.333, 25.38, 25.04, 97.54],
    [4, 0.2, 9.08, 1.79, 69.20],
    [6, 0.5, 24.03, 24.14, 98.71],
    [6, 0.333, 9.65, 1.75, 75.99],
    [6, 0.2, 4.99, 0.23, 48.12],
]

df = pd.DataFrame(data, columns=["Médicos", "TasaLlegada", "EsperaProm", "TasaAbandono", "UsoMedico"])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for tasa in df["TasaLlegada"].unique():
    subset = df[df["TasaLlegada"] == tasa]
    plt.plot(subset["Médicos"], subset["EsperaProm"], marker='o', label=f"Tasa llegada = {tasa:.3f}")

plt.title("Relación entre número de médicos y espera promedio")
plt.xlabel("Número de médicos")
plt.ylabel("Espera promedio (min)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for tasa in df["TasaLlegada"].unique():
    subset = df[df["TasaLlegada"] == tasa]
    plt.plot(subset["Médicos"], subset["TasaAbandono"], marker='o', label=f"Tasa llegada = {tasa:.3f}")

plt.title("Relación entre número de médicos y tasa de abandono")
plt.xlabel("Número de médicos")
plt.ylabel("Tasa de abandono (%)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for tasa in df["TasaLlegada"].unique():
    subset = df[df["TasaLlegada"] == tasa]
    plt.plot(subset["Médicos"], subset["UsoMedico"], marker='o', label=f"Tasa llegada = {tasa:.3f}")

plt.title("Relación entre número de médicos y uso promedio")
plt.xlabel("Número de médicos")
plt.ylabel("Uso promedio de médicos (%)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(3, 1, figsize=(10, 15), sharex=True)

metricas = ["EsperaProm", "TasaAbandono", "UsoMedico"]
titulos = ["Espera promedio (min)", "Tasa de abandono (%)", "Uso de médicos (%)"]

for i, metrica in enumerate(metricas):
    for tasa in df["TasaLlegada"].unique():
        subset = df[df["TasaLlegada"] == tasa]
        axs[i].plot(subset["Médicos"], subset[metrica], marker='o', label=f"Tasa llegada = {tasa:.3f}")
    axs[i].set_ylabel(titulos[i])
    axs[i].legend()
    axs[i].grid(True)

axs[2].set_xlabel("Número de médicos")
fig.suptitle("Efecto del número de médicos en el desempeño del sistema", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["EsperaProm"], df["TasaAbandono"], color='b', label="Tasa de abandono vs Tiempo promedio de espera")
plt.title("Relación entre tiempo promedio de espera y tasa de abandono")
plt.xlabel("Tiempo promedio de espera (min)")
plt.ylabel("Tasa de abandono (%)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
